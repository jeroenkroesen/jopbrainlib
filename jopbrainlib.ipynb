# JopBrainLib

What is the default template folder and filenames?
Title template: diary_day_title.md
Body template: diary_day_body.md

Folder: Path(__file__).parent.joinpath('templates')
(So it's relative to the installed path)
Folder arg is a string, turned into a Path that is fully resolved so ~ is supported

Where are the templates according to my stow?
~/.config/jopbrainlib/my_templates

```python
from joppylib import default_settings, JoplinClient
from jopbrainlib.diary import create_diary_day, create_diary_month

template_folder = '/home/jeroen/.config/jopbrainlib/my_templates'

joplin = JoplinClient(default_settings)
```
